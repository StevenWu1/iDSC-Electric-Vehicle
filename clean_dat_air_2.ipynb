```python
import pandas as pd
```

```python
air_quality_2 = pd.read_csv("PM25_C1.csv")
air_quality_2 = air_quality_2.drop(columns=["Unnamed: 0", "index"])
air_quality_2.head(5)
```

```python
cities = air_quality_2['City or Locality'].unique()
cities
# cities = ['Adelaide']
```

```python
df_test = pd.DataFrame(data={"WHO Country Name": [], "City or Locality": [], "Measurement Year": [], "PM2.5 (μg/m3)": [], "PM25 temporal coverage (%)": []})
for city in cities:
    df_current_city = air_quality_2[air_quality_2['City or Locality'] == city]
    replace_value = df_current_city['PM25 temporal coverage (%)'].mean()
    df_current_city = df_current_city.fillna(value = float(replace_value))
    df_test = pd.concat([df_test, df_current_city])
```

```python
df_test["PM25 temporal coverage (%)"] = df_test["PM25 temporal coverage (%)"].astype(float).round(2)
df_test.head(10)
```

```python
df_test.to_csv("clean_test.csv")
```
